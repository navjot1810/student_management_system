```python
import tkinter as tk
from tkinter import ttk, messagebox


class StudentManagement:
    def __init__(self, root):
        self.root = root
        self.root.title("Student Result Management System")
        self.root.geometry("1100x650")
        self.root.resizable(False, False)
        blue = "#06466B"
        light_yellow = "#FFFFE0"

        title = tk.Label(
            root,
            text="Manage Student Details",
            font=("Arial", 20, "bold"),
            bg=blue,
            fg="white",
            pady=8
        )
        title.pack(fill="x", padx=10, pady=(10, 5))

        main_frame = tk.Frame(root)
        main_frame.pack(fill="both", expand=True, padx=15)

        left_frame = tk.Frame(main_frame)
        left_frame.place(x=0, y=10, width=620, height=550)

        label_font = ("Arial", 12, "bold")
        entry_font = ("Arial", 12)

        tk.Label(left_frame, text="Roll No.", font=label_font).place(x=0, y=10)
        self.roll_var = tk.StringVar()
        tk.Entry(
            left_frame,
            textvariable=self.roll_var,
            font=entry_font,
            bg=light_yellow
        ).place(x=125, y=8, width=180, height=27)

        # DOB
        tk.Label(left_frame, text="D.O.B.", font=label_font).place(x=315, y=10)
        self.dob_var = tk.StringVar()
        tk.Entry(
            left_frame,
            textvariable=self.dob_var,
            font=entry_font,
            bg=light_yellow
        ).place(x=420, y=8, width=180, height=27)

        # Name
        tk.Label(left_frame, text="Name", font=label_font).place(x=0, y=46)
        self.name_var = tk.StringVar()
        tk.Entry(
            left_frame,
            textvariable=self.name_var,
            font=entry_font,
            bg=light_yellow
        ).place(x=125, y=44, width=180, height=27)

        # Contact
        tk.Label(left_frame, text="Contact", font=label_font).place(x=315, y=46)
        self.contact_var = tk.StringVar()
        tk.Entry(
            left_frame,
            textvariable=self.contact_var,
            font=entry_font,
            bg=light_yellow
        ).place(x=420, y=44, width=180, height=27)

        # Email
        tk.Label(left_frame, text="Email", font=label_font).place(x=0, y=82)
        self.email_var = tk.StringVar()
        tk.Entry(
            left_frame,
            textvariable=self.email_var,
            font=entry_font,
            bg=light_yellow
        ).place(x=125, y=80, width=180, height=27)

        # Admission
        tk.Label(left_frame, text="Admission", font=label_font).place(x=315, y=82)
        self.admission_var = tk.StringVar()
        tk.Entry(
            left_frame,
            textvariable=self.admission_var,
            font=entry_font,
            bg=light_yellow
        ).place(x=420, y=80, width=180, height=27)

        # Gender
        tk.Label(left_frame, text="Gender", font=label_font).place(x=0, y=118)

        self.gender_var = tk.StringVar(value="Select")
        gender_combo = ttk.Combobox(
            left_frame,
            textvariable=self.gender_var,
            values=["Select", "Male", "Female", "Other"],
            state="readonly",
            font=entry_font
        )
        gender_combo.place(x=125, y=116, width=180, height=28)

        # State
        tk.Label(left_frame, text="State", font=label_font).place(x=0, y=155)

        self.state_var = tk.StringVar()
        tk.Entry(
            left_frame,
            textvariable=self.state_var,
            font=entry_font,
            bg=light_yellow
        ).place(x=125, y=153, width=135, height=27)

        # City
        tk.Label(left_frame, text="City", font=label_font).place(x=275, y=155)

        self.city_var = tk.StringVar()
        tk.Entry(
            left_frame,
            textvariable=self.city_var,
            font=entry_font,
            bg=light_yellow
        ).place(x=330, y=153, width=100, height=27)

        # Pin
        tk.Label(left_frame, text="Pin", font=label_font).place(x=445, y=155)

        self.pin_var = tk.StringVar()
        tk.Entry(
            left_frame,
            textvariable=self.pin_var,
            font=entry_font,
            bg=light_yellow
        ).place(x=485, y=153, width=115, height=27)

        # Address
        tk.Label(left_frame, text="Address", font=label_font).place(x=0, y=195)

        self.address_text = tk.Text(
            left_frame,
            font=("Arial", 12),
            bg=light_yellow
        )
        self.address_text.place(x=125, y=190, width=475, height=110)

        # ================= BUTTONS =================

        tk.Button(
            left_frame,
            text="Save",
            font=("Arial", 12, "bold"),
            bg="#2196F3",
            fg="white",
            width=9,
            command=self.save_student
        ).place(x=125, y=325)

        tk.Button(
            left_frame,
            text="Update",
            font=("Arial", 12, "bold"),
            bg="#4CAF50",
            fg="white",
            width=9,
            command=self.update_student
        ).place(x=235, y=325)

        tk.Button(
            left_frame,
            text="Delete",
            font=("Arial", 12, "bold"),
            bg="#F44336",
            fg="white",
            width=9,
            command=self.delete_student
        ).place(x=345, y=325)

        tk.Button(
            left_frame,
            text="Clear",
            font=("Arial", 12, "bold"),
            bg="#607D8B",
            fg="white",
            width=9,
            command=self.clear_fields
        ).place(x=455, y=325)

      
        right_frame = tk.Frame(main_frame)
        right_frame.place(x=635, y=10, width=430, height=550)

        # Search
        tk.Label(
            right_frame,
            text="Roll No.",
            font=label_font
        ).place(x=0, y=10)

        self.search_var = tk.StringVar()

        tk.Entry(
            right_frame,
            textvariable=self.search_var,
            font=entry_font,
            bg=light_yellow
        ).place(x=130, y=8, width=165, height=27)

        tk.Button(
            right_frame,
            text="Search",
            font=("Arial", 11, "bold"),
            bg="#03A9F4",
            fg="white",
            command=self.search_student
        ).place(x=305, y=7, width=105, height=29)

        # Available Courses
        tk.Label(
            right_frame,
            text="Available Courses",
            font=label_font
        ).place(x=0, y=48)

        self.course_var = tk.StringVar()

        self.course_combo = ttk.Combobox(
            right_frame,
            textvariable=self.course_var,
            values=[
                "Python",
                "Java",
                "C++",
                "Data Science",
                "Web Development",
                "Database Management",
                "Artificial Intelligence"
            ],
            state="readonly"
        )
        self.course_combo.place(x=0, y=78, width=215, height=28)

        tk.Button(
            right_frame,
            text="Add Course →",
            font=("Arial", 10, "bold"),
            bg="#2196F3",
            fg="white",
            command=self.add_course
        ).place(x=230, y=77, width=110, height=28)

        tk.Button(
            right_frame,
            text="← Remove",
            font=("Arial", 10, "bold"),
            bg="#F44336",
            fg="white",
            command=self.remove_course
        ).place(x=230, y=112, width=110, height=28)

        # Enrolled Courses
        tk.Label(
            right_frame,
            text="Enrolled Courses",
            font=label_font
        ).place(x=0, y=145)

        self.course_list = tk.Listbox(
            right_frame,
            font=("Arial", 11),
            selectmode=tk.SINGLE
        )
        self.course_list.place(x=0, y=175, width=340, height=120)

        # ==================================================
        # STUDENT TABLE
        # ==================================================

        table_frame = tk.Frame(right_frame)
        table_frame.place(x=0, y=320, width=410, height=190)

        columns = ("roll", "name", "course")

        self.student_table = ttk.Treeview(
            table_frame,
            columns=columns,
            show="headings"
        )

        self.student_table.heading("roll", text="Roll No.")
        self.student_table.heading("name", text="Name")
        self.student_table.heading("course", text="Course(s)")

        self.student_table.column("roll", width=80)
        self.student_table.column("name", width=110)
        self.student_table.column("course", width=190)

        scrollbar = ttk.Scrollbar(
            table_frame,
            orient="vertical",
            command=self.student_table.yview
        )

        self.student_table.configure(
            yscrollcommand=scrollbar.set
        )

        self.student_table.pack(side="left", fill="both", expand=True)
        scrollbar.pack(side="right", fill="y")

        self.student_table.bind(
            "<ButtonRelease-1>",
            self.get_selected_student
        )

    # ==================================================
    # FUNCTIONS
    # ==================================================

    def add_course(self):
        course = self.course_var.get()

        if course == "":
            messagebox.showwarning(
                "Warning",
                "Please select a course."
            )
            return

        courses = self.course_list.get(0, tk.END)

        if course not in courses:
            self.course_list.insert(tk.END, course)

    def remove_course(self):
        selected = self.course_list.curselection()

        if selected:
            self.course_list.delete(selected)

    def get_courses(self):
        return ", ".join(self.course_list.get(0, tk.END))

    def save_student(self):
        roll = self.roll_var.get()
        name = self.name_var.get()
        courses = self.get_courses()

        if roll == "" or name == "":
            messagebox.showerror(
                "Error",
                "Roll No. and Name are required."
            )
            return

        self.student_table.insert(
            "",
            tk.END,
            values=(roll, name, courses)
        )

        messagebox.showinfo(
            "Success",
            "Student saved successfully!"
        )

    def update_student(self):
        selected = self.student_table.selection()

        if not selected:
            messagebox.showwarning(
                "Warning",
                "Please select a student from the table."
            )
            return

        courses = self.get_courses()

        self.student_table.item(
            selected[0],
            values=(
                self.roll_var.get(),
                self.name_var.get(),
                courses
            )
        )

        messagebox.showinfo(
            "Success",
            "Student updated successfully!"
        )

    def delete_student(self):
        selected = self.student_table.selection()

        if not selected:
            messagebox.showwarning(
                "Warning",
                "Please select a student."
            )
            return

        answer = messagebox.askyesno(
            "Delete",
            "Are you sure you want to delete this student?"
        )

        if answer:
            self.student_table.delete(selected[0])
            self.clear_fields()

    def search_student(self):
        roll = self.search_var.get()

        for item in self.student_table.get_children():
            values = self.student_table.item(item)["values"]

            if str(values[0]) == roll:
                self.student_table.selection_set(item)
                self.student_table.focus(item)
                self.get_selected_student()
                return

        messagebox.showinfo(
            "Search",
            "Student not found."
        )

    def get_selected_student(self, event=None):
        selected = self.student_table.selection()

        if not selected:
            return

        values = self.student_table.item(
            selected[0]
        )["values"]

        self.roll_var.set(values[0])
        self.name_var.set(values[1])

        # Clear previous courses
        self.course_list.delete(0, tk.END)

        if values[2]:
            courses = str(values[2]).split(", ")

            for course in courses:
                self.course_list.insert(
                    tk.END,
                    course
                )

    def clear_fields(self):
        self.roll_var.set("")
        self.name_var.set("")
        self.dob_var.set("")
        self.contact_var.set("")
        self.email_var.set("")
        self.admission_var.set("")
        self.gender_var.set("Select")
        self.state_var.set("")
        self.city_var.set("")
        self.pin_var.set("")
        self.search_var.set("")
        self.course_var.set("")

        self.address_text.delete(
            "1.0",
            tk.END
        )

        self.course_list.delete(
            0,
            tk.END
        )

        self.student_table.selection_remove(
            self.student_table.selection()
        )


# ================= RUN APPLICATION =================

if __name__ == "__main__":
    root = tk.Tk()
    app = StudentManagement(root)
    root.mainloop()
```
